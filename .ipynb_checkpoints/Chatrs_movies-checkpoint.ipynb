```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from utils import load_clean_data 
pd.set_option('display.max_rows', None)
pd.set_option('display.width', None)
```

```python
movies = load_clean_data()
print(movies.head())
```

## Histogram of vote_avarage and revenue

### Vote_avarage

```python
custom_bins = np.arange(0,10.1,0.5)

plt.figure(figsize=(8,6))

plt.hist(movies['vote_average'], bins = custom_bins, edgecolor = 'black')
plt.title('Histogram of vote_average')
plt.xlabel('vote_average')
plt.ylabel('number of votes')
plt.xticks(np.arange(0,10.1,0.5))

plt.show()
```

### Revenue

```python
plt.figure(figsize=(8,6))

amounts = [1e6, 5e6, 1e7, 5e7, 1e8, 2e8, 5e8, 1e9, 2e9]
positions = np.log(amounts)
labels = ['1M$','5M$','10M$','50M$','100M$','200M$','500M$','1B$','2B$']

plt.hist(np.log(movies['revenue']), bins = 50, edgecolor = 'black')
plt.xticks(positions, labels, rotation = 90)
plt.ylabel('number_of_movies')
plt.xlabel('log(revenue)')
plt.title('Distribution of film revenues')

plt.show()
```
