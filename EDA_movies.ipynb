```python
import pandas as pd
import numpy as np
from utils import load_clean_data 
pd.set_option('display.max_rows', None)
pd.set_option('display.width', None)
```

```python
movies = load_clean_data()
print(movies.head())
```

```python
skosnosc = movies[['budget','revenue','runtime','vote_count','popularity']]
print(skosnosc.skew(numeric_only=True))
```

## Liczenie po gatunkach

```python
genres = movies.explode('genre_list')
print(genres['genre_list'].value_counts())
```

## Dekady

```python
dekady = range(1910,2030,10)
etykiety = [f"{rok}s" for rok in dekady[:-1]]

count_decades = pd.cut(movies['release_year'], bins = dekady, labels=etykiety).value_counts().sort_index(ascending = False)
print(count_decades)
```
